## Choose your adapter

```python
# Set this to False if you didn't save your own adapter in Chapter 5.
USE_MY_ADAPTER = True

MY_ADAPTER = "/content/drive/MyDrive/ridgeline-lora"
INSTRUCTOR_ADAPTER = "KWillz/ridgeline-support-lora"

if USE_MY_ADAPTER:
    from google.colab import drive
    drive.mount("/content/drive")
    adapter = MY_ADAPTER
else:
    adapter = INSTRUCTOR_ADAPTER
print(f"Using adapter: {adapter}")
```

## Load the fine-tuned model

```python
from unsloth import FastLanguageModel

# Loading an adapter also loads the base model it was trained on.
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=adapter,
    max_seq_length=2048,
    load_in_4bit=True,
)
model = FastLanguageModel.for_inference(model)
```

## Load the evaluation tickets

```python
eval_tickets = ridgeline.load_eval_tickets()
print(f"{len(eval_tickets)} evaluation tickets. The model never trained on any of them.")
```

## Run both models on every ticket

`disable_adapter` turns the LoRA weights off, so one loaded model gives you both the baseline and the fine-tuned results. Each model gets the prompt it would actually use in production. This takes several minutes.

```python
import time

outputs = []
start = time.time()
for i, ticket in enumerate(eval_tickets, start=1):
    # Baseline: the base model, with the full voice rules in the prompt.
    with model.disable_adapter():
        baseline = ridgeline.generate(model, tokenizer, ridgeline.build_messages(ticket, include_voice_rules=True))
    # Fine-tuned: the adapter on, with no voice rules in the prompt.
    finetuned = ridgeline.generate(model, tokenizer, ridgeline.build_messages(ticket, include_voice_rules=False))
    outputs.append({"id": ticket["id"], "baseline": baseline, "finetuned": finetuned})
    if i % 8 == 0:
        print(f"  {i}/{len(eval_tickets)} tickets done ({(time.time() - start) / 60:.1f} min)")
```

## Score both models

```python
import pandas as pd

def summarize(model_name):
    results = [ridgeline.check_reply(t, o[model_name]) for t, o in zip(eval_tickets, outputs)]
    header = "missing or broken Category/Escalate header"
    return {
        "Category correct": sum(r["category_ok"] for r in results) / len(results),
        "Escalate correct": sum(r["escalate_ok"] for r in results) / len(results),
        "Header format correct": sum(header not in r["failures"] for r in results) / len(results),
        "Passed every voice rule": sum(r["voice_ok"] for r in results) / len(results),
    }

table = pd.DataFrame({"Baseline": summarize("baseline"), "Fine-tuned": summarize("finetuned")})
table.style.format("{:.0%}")
```

## Read the category misses

A percentage hides the details. Print every ticket where the fine-tuned model chose a different category than the label, and read the subjects. Some misses are real mistakes. Others are tickets where the label itself is debatable.

```python
for t, o in zip(eval_tickets, outputs):
    got = ridgeline.parse_output(o["finetuned"])["category"]
    if got != t["category"]:
        print(f"{t['id']:9} expected {t['category']:20} got {got}  | {t['subject'][:50]}")
```

## Compare prompt sizes

```python
prompt_sizes = {
    "Baseline": ridgeline.count_tokens(tokenizer, ridgeline.build_messages(eval_tickets[0], True)),
    "Fine-tuned": ridgeline.count_tokens(tokenizer, ridgeline.build_messages(eval_tickets[0], False)),
}
for name, size in prompt_sizes.items():
    print(f"{name} prompt: {size} tokens per ticket")
```

## Look at one ticket side by side

```python
ticket = next(t for t in eval_tickets if t["id"] == "RT-008")
pair = next(o for o in outputs if o["id"] == "RT-008")
print(f"{ticket['subject']}\n")
print("BASELINE\n" + pair["baseline"] + "\n")
print("FINE-TUNED\n" + pair["finetuned"])
```
